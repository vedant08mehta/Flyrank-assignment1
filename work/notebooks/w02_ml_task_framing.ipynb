## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Classification. I want to classify content pages according to whether their search performance is declining. Classification fits because each content item can be assigned a binary outcome, such as declining or not declining, which can then be used to prioritize pages for review.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("Task type: binary classification")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is whether a content page is showing a declining trend. I will use the observed trend_direction field and define the target as trend_direction = "down". This is a defined label based on an observed outcome in the dataset, rather than something that I will claim represents Google's algorithm.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df["is_declining_label"] = (
    df["trend_direction"] == "down"
).astype(int)

print(df["is_declining_label"].value_counts())
print(f"Declining rate: {df['is_declining_label'].mean():.3f}")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

A good model should place genuinely declining pages near the top of the ranked list. I would consider the model useful if Precision@50 is meaningfully better than a simple baseline, while treating the exact threshold as dependent on the available review capacity.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("Primary metric: Precision@50")
print("Goal: rank genuinely declining pages near the top of the review queue.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule based on a single metric, such as impressions or CTR, may miss pages that show more complex patterns of underperformance. An ML model can combine multiple signals, such as search performance, engagement, content age, and position, to identify patterns that are harder to capture with one threshold. This could help prioritize pages for review more efficiently. However, the model would be used for decision support and prioritization, not as proof that any particular factor causes performance changes.
