## 1. Question

*The research question and the decision it supports.*

Question: Can we use the information we already have about a content page to figure out which pages are more likely to be declining and should be looked at first?

The main idea is to help the content team decide where to spend their time. Instead of going through thousands of pages one by one, the model can give us a ranked list of pages that are worth checking.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

I used the anonymized content-refresh dataset provided for the project. It has 30,000 rows and 45 columns, with each row representing one content page.

The data includes things like search volume, impressions, clicks, CTR, average position, engagement, content age, freshness, and recent traffic trends. I used the available 90-day and 30-day data windows. Since this is supposed to be public-safe, I did not use any client names, URLs, or private search queries.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

I treated this as a classification problem where the goal is to identify pages that are showing a declining trend. The label comes from the observed trend_direction field.

I used the search, traffic, engagement, content length, age, and freshness information as features. I also checked the features for possible leakage and avoided using information that would effectively give the model the answer beforehand.

For evaluation, I used Precision@50 because the main goal is to get the most useful pages into the top of the review list. I compared the model against the baseline using the same held-out data so that the comparison was fair.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

The model did better than the baseline on the honest test split. The model got a Precision@50 of 0.680, while the baseline got 0.380.

That is an improvement of 0.300 in Precision@50. The test set had a base rate of 0.517, so the model was able to put more of the declining pages towards the top of the list instead of just selecting pages at roughly the overall rate.

This makes the model useful for deciding which pages should be checked first, although it doesn't mean every page it recommends definitely needs to be changed.

## 5. Limitations

*What this work cannot claim.*

There are a few things this model can't tell us. For example, just because a page has a certain feature and is declining doesn't mean that feature is actually causing the decline.

The 0.680 Precision@50 result is also based on this particular dataset and test split. Performance could be different with new data or if search behaviour and traffic patterns change.

So I would use the model to make a shortlist of pages to investigate, rather than letting it automatically decide which pages should be changed.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

The ranked list is basically a way of deciding what to look at first. Pages near the top have stronger signals that make them worth investigating.

A high ranking doesn't automatically mean that the page should be refreshed. Someone from the content team should still look at the page, its search intent, recent changes, and the surrounding context before deciding what to do.
